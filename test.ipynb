```python
from paptools import *


n1 = Array([1.367, 2.345, 3.345, 4.345, 5.345], .123, symbol="n_1")

t1 = Array([10, 20, 30, 40, 50], symbol="t_1", unit="s")

r1 = (n1 / t1).round()

print(r1)

num_messung = Array([1, 2, 3, 4, 5], symbol="Messung")
error = r1._error
value = r1._value

#plot with error bars
import matplotlib.pyplot as plt
plt.errorbar(num_messung._value, value, yerr=error, fmt='o', label='Messung mit Fehlerbalken')
plt.xlabel('Messung')
plt.ylabel('Wert')
plt.title('Messung mit Fehlerbalken')
plt.legend()
plt.show()
```
